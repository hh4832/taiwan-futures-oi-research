# 02 — Validation

Checks temporal alignment, missing tails, infinities, and basic group behavior.

```
import numpy as np
import pandas as pd

daily = pd.read_parquet("outputs/daily_dataset.parquet")
results = pd.read_csv("outputs/primary_results.csv")

print(daily.shape, results.shape)
display(results.head())
```

```
numeric = daily.select_dtypes(include="number")
assert not np.isinf(numeric.to_numpy()).any(), "Infinity detected"
print("PASS: no +/- infinity in numeric columns")
```

```
targets = [c for c in daily.columns if "fwd_" in c]
for c in targets:
    print(c, "tail NaN =", bool(pd.isna(daily[c].iloc[-1])))
```
